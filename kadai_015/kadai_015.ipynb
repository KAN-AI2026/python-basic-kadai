```python
# 1. Humanクラスを作成する
class Human:
    # コンストラクタ（初期化メソッド）：nameとageを属性として持たせる
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 2. 標準出力するメソッド(printinfo)を追加する
    def printinfo(self):
        # 名前と年齢を出力
        print(f"名前: {self.name}")
        print(f"年齢: {self.age}")


# 3. Humanクラスのインスタンスを生成し、変数に代入する
# （名前と年齢は好きな値を入れてOKです。ここでは「侍太郎」「35」とします）
person = Human("侍太郎", 35)

# 4. 定義したメソッド(printinfo)を実行する
person.printinfo()
```

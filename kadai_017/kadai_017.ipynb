```python
# 1. Humanクラスを作成
class Human:
    # コンストラクタ（名前と年齢を属性として持たせる）
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # check_adultメソッドを追加
    def check_adult(self):
        # 条件分岐：ageが20以上の場合
        if self.age >= 20:
            print(f"{self.name}は大人である")
        # そうでない場合
        else:
            print(f"{self.name}は大人でない")


# 2. 複数のインスタンスを生成してリストに追加する
human1 = Human("侍太郎", 25)  # 25歳（大人）
human2 = Human("侍花子", 18)  # 18歳（未成年）
human3 = Human("侍次郎", 20)  # 20歳（大人）

# インスタンスをリストにまとめる
human_list = [human1, human2, human3]

# 3. リストの要素数分だけcheck_adultメソッドを呼び出す
for human in human_list:
    human.check_adult()
```
